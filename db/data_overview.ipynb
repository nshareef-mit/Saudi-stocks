```python
import pandas as pd
import numpy as np
import plotly.express as px
import seaborn as sns
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8")
```

## Database Connection

```python
import os
import psycopg2
import pandas as pd
from dotenv import load_dotenv

# Load .env
load_dotenv()

conn = psycopg2.connect(
    dbname=os.getenv("DB_NAME"),
    user=os.getenv("DB_USER"),
    password=os.getenv("DB_PASSWORD"),
    host=os.getenv("DB_HOST"),
    port=os.getenv("DB_PORT"),
)
```

## DataVisualizer Class

```python
class DataVisualizer:

    def __init__(self, conn):
        self.conn = conn

    def load_table(self, table_name):
        return pd.read_sql(f"SELECT * FROM {table_name}", self.conn)

    # -----------------------------
    # NEWS ANALYSIS
    # -----------------------------
    def analyze_news(self):
        df = self.load_table("news")

        print("Total News:", len(df))
        print("Earliest Date:", df["published_date"].min())

        df["year"] = pd.to_datetime(df["published_date"]).dt.year

        yearly = df.groupby("year").size().reset_index(name="count")

        fig = px.bar(
            yearly,
            x="year",
            y="count",
            title="News Count Per Year",
            text_auto=True,
        )
        fig.show()

        source_dist = df["source"].value_counts().reset_index()
        source_dist.columns = ["source", "count"]

        fig2 = px.pie(
            source_dist,
            names="source",
            values="count",
            title="News Distribution by Source",
        )
        fig2.show()

    # -----------------------------
    # ANNOUNCEMENTS
    # -----------------------------
    def analyze_announcements(self):
        df = self.load_table("announcements")

        print("Total Announcements:", len(df))
        print("Earliest Date:", df["announcement_date"].min())

        df["year"] = pd.to_datetime(df["announcement_date"]).dt.year

        yearly = df.groupby("year").size().reset_index(name="count")

        fig = px.bar(
            yearly,
            x="year",
            y="count",
            title="Announcements Per Year",
            text_auto=True,
        )
        fig.show()

    # -----------------------------
    # SOCIAL POSTS
    # -----------------------------
    def analyze_social(self):
        df = self.load_table("social_posts")

        print("Total Posts:", len(df))
        print("Earliest Date:", df["published_date"].min())

        df["year"] = pd.to_datetime(df["published_date"]).dt.year

        yearly = df.groupby("year").size().reset_index(name="count")

        fig = px.bar(
            yearly,
            x="year",
            y="count",
            title="Social Posts Per Year",
            text_auto=True,
        )
        fig.show()

        platform_dist = df["platform"].value_counts().reset_index()
        platform_dist.columns = ["platform", "count"]

        fig2 = px.pie(
            platform_dist,
            names="platform",
            values="count",
            title="Posts by Platform",
        )
        fig2.show()

    # -----------------------------
    # PRICES
    # -----------------------------
    def analyze_prices(self):
        df = self.load_table("prices")
        print("Total Price Records:", len(df))

    # -----------------------------
    # FINANCIALS
    # -----------------------------
    def analyze_financials(self):
        df = self.load_table("financials")
        print("Total Financial Records:", len(df))

    # -----------------------------
    # OVERVIEW
    # -----------------------------
    def overview(self):
        tables = [
            "companies",
            "prices",
            "daily_prices",
            "financials",
            "dividends",
            "announcements",
            "news",
            "social_posts",
        ]

        counts = []

        for table in tables:
            df = self.load_table(table)
            counts.append({"table": table, "rows": len(df)})

        df_counts = pd.DataFrame(counts)

        fig = px.bar(
            df_counts,
            x="table",
            y="rows",
            title="Database Tables Overview",
            text_auto=True,
        )
        fig.show()
```

## overview 

```python
import psycopg2

conn = psycopg2.connect(
    dbname=os.getenv("DB_NAME"),
    user=os.getenv("DB_USER"),
    password=os.getenv("DB_PASSWORD"),
    host=os.getenv("DB_HOST"),
    port=os.getenv("DB_PORT"),
)

viz = DataVisualizer(conn)


viz.analyze_news()
viz.analyze_announcements()

conn.close()
```
