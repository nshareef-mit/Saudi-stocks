# Data preperation

## Training Data (X) and Target Data (y) — Column Breakdown

The key insight is that both X and y must share **the same unit: one row per trading day**. Here is exactly what each stage produces.

---

### Target Data (y) — From the JSON File

First, note that your JSON doesn't have an explicit `closePrice` column. But `previousClosePrice` on day D is actually the close of day D-1, so the daily return for day D is:

$$\text{return}_D = \frac{\text{prevClose}_{D+1} - \text{prevClose}_D}{\text{prevClose}_D} \times 100$$

| column | source | example |
|---|---|---|
| `date` | `transactionDate` | 2026-04-07 |
| `prev_close` | `previousClosePrice` | 11,087.54 |
| `tasi_return` ✅ **this is y** | computed from above formula | +2.27% |

---

### Aggregation Strategy before talking about Training Data (X) 
### Option A — Mean Embedding Per Day (Simple + Good)

For each day:

\[
X_t = \frac{1}{N_t} \sum_{i=1}^{N_t} embedding_i
\]

This gives:

- One 1536‑dimensional vector per day
- Matches your y dimensionally

This is the cleanest first regression.

---

### Option B — News Volume + Mean Embedding

Add:

- Number of news that day
- Maybe number of announcements

So X becomes:

```
[mean_embedding_1536 , news_count]
```

This often improves performance.

---

### Option C — Sentiment Projection (More Advanced)

Instead of raw 1536 dims, project onto:

- Positive direction
- Negative direction
- Topic clusters

But we don’t start there.

---

Option 2 — Reduce dimensions (recommended later)

This:

Reduces noise
Improves stability
Speeds training


**Note**
File: aggregate_daily_embeddings.py

This script pulls all news and announcement embeddings from the database, groups them by news_date, and computes the mean embedding per day to create one semantic vector per trading day (809 days × 1536 features).
It then applies PCA (50 components) to reduce dimensionality and stabilize regression.
The 50 principal components capture ~78% of the total embedding variance, meaning most daily narrative information is preserved while significantly reducing model complexity.

SO we have : X = daily semantic market narrative

## Estimation by Ridge Regression:

- **X → Ridge Regression → β**

 What This Does

Mathematically:

\[
\hat{y} = X \beta
\]

Where:

- \( X \) = PCA components
- \( \beta \) = learned weights
- Ridge adds L2 regularization:

\[
\min_{\beta} ||y - X\beta||^2 + \alpha ||\beta||^2
\]

## Next-Day Prediction Setup

Instead of predicting same-day returns, we shift the target forward:

- \( X_t \) = daily semantic narrative (PCA embeddings on day t)  
- \( y_{t+1} \) = TASI return on the next trading day  

This tests whether today’s news explains **tomorrow’s market movement**, which is more realistic since markets may digest information with delay.
